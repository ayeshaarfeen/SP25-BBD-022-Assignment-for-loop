```python
#question6
# Products: name -> [price, quantity]
products = {
    "Book": [10.0, 100],
    "Headphones": [50.0, 20],
    "Mouse": [25.0, 15]
}

# Shopping cart: name -> quantity
cart = {}

def add_to_cart(name, qty):
    if name in products and products[name][1] >= qty:
        cart[name] = cart.get(name, 0) + qty
        products[name][1] -= qty

def remove_from_cart(name, qty):
    if name in cart:
        cart[name] -= qty
        products[name][1] += qty
        if cart[name] <= 0:
            del cart[name]

def calculate_total(discount=0.1, tax=0.07):
    subtotal = sum(products[item][0] * qty for item, qty in cart.items())
    total = subtotal - (subtotal * discount) + (subtotal * tax)
    return round(total, 2)

# Example usage
add_to_cart("Book", 2)
add_to_cart("Mouse", 1)

print("Cart:", cart)
print("Total (with discount & tax): $", calculate_total())
```

```python
#question7
# Dictionary: student -> {subject: grade}
gradebook = {
    "Alice": {"Math": 90, "Science": 85},
    "Bob": {"Math": 70, "Science": 80},
    "Charlie": {"Math": 95, "Science": 88}
}

# Calculate average grade per student
for student, subjects in gradebook.items():
    avg = sum(subjects.values()) / len(subjects)
    print(f"{student}'s average: {avg}")

# Highest and lowest grade in Math
math_grades = {name: info["Math"] for name, info in gradebook.items()}
highest = max(math_grades.items(), key=lambda x: x[1])
lowest = min(math_grades.items(), key=lambda x: x[1])

print("Highest in Math:", highest)
print("Lowest in Math:", lowest)

# Sort students by average grade
sorted_students = sorted(gradebook.items(), key=lambda x: sum(x[1].values()) / len(x[1]), reverse=True)
print("Students sorted by average grade:")
for name, subjects in sorted_students:
    avg = sum(subjects.values()) / len(subjects)
    print(f"{name}: {avg}")
```
