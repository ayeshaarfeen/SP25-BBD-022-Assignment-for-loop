```python
#Q5
def get_formatted_name(first, last):
    full_name= f"{first} {last}"
    return full_name.title()
print(get_formatted_name('Ayesha' , 'Arfeen'))
```

```python
#Q6
def get_formatted_name(first, last, middle= ''):
    if middle:
        full_name = f"{first} {middle} {last}"
    else:
            full_name = f"{first} {last}"
    return full_name.title()
print (get_formatted_name('Ayesha' , 'Arfeen'))
print(get_formatted_name('Ayesha' , 'Arfeen' , 'Bokhari'))
```

```python
#Q7
def build_person(first, last, age=None):
    return {'first': first, 'last': last, 'age':age}
print(build_person('Ayesha', 'Arfeen'))
print(build_person('Ayesha', 'Arfeen', 18))
```

```python
#Q8
def greet_users(usernames):
    for username in usernames:
        print(f"Hello, {username}!")
greet_users(['Alice', 'Bob', 'Hiba'])
```

```python
#Q9
def make_pizza(*toppings):
    print("\nMaking a pizza with the following toppings:")
    for topping in toppings:
        print(f"- {topping}")

make_pizza('pepperoni', 'mushrooms', 'extra cheese')
```

```python
#Q10
def build_profile(first, last, **info):
    profile = {'first': first, 'last': last}
    for key, value in info.items():
        profile[key] = value
    return profile

print(build_profile('John', 'Doe', location='USA', profession='Engineer'))
```
