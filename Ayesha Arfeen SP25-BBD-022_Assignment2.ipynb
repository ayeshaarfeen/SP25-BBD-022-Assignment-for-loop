```python
# Ask the user to enter a number
number = int(input("Enter a number: "))

# Check divisibility and print appropriate output
if number % 3 == 0 and number % 5 == 0:
    print("FizzBuzz")
elif number % 3 == 0:
    print("Fizz")
elif number % 5 == 0:
    print("Buzz")
else:
    print(number)
```

```python
# Ask the user to enter a character
char = input("Enter a single character: ")

# Check input validity
if len(char) != 1:
    print("Please enter exactly one character.")
else:
    if char.isalpha():
        if char.lower() in 'aeiou':
            print("Vowel")
        else:
            print("Consonant")
    elif char.isdigit():
        print("Digit")
    else:
        print("Special character")
```

```python
# Ask the user to enter the current day name
day = input("Enter the current day (e.g., Monday, Tuesday): ")

# Normalize input to capitalize the first letter (e.g., "monday" -> "Monday")
day = day.capitalize()

# Check if it's a weekend or a weekday
if day == "Saturday" or day == "Sunday":
    print("Enjoy your weekend!")
else:
    print("Time to work!")
```

```python
# Get user input for years of service and salary
years = int(input("Enter your years of service: "))
salary = float(input("Enter your current salary: "))

# Determine bonus percentage based on years of service
if years >= 10:
    bonus_percentage = 0.20
elif years >= 5:
    bonus_percentage = 0.10
else:
    bonus_percentage = 0.05

# Calculate bonus amount
bonus = salary * bonus_percentage

# Print the bonus amount
print(f"Your bonus is: ${bonus:.2f}")
```
